Sub plot

```python
# First graph
plt.subplot(1, 2, 1)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.title("Line Chart")


# Second graph
plt.subplot(1, 2, 2)

subjects = ["Python", "Java", "SQL"]
marks = [80, 70, 90]

plt.bar(subjects, marks)
plt.title("Bar Chart")

plt.show()
```

```python


x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

subjects = ["Python", "Java", "SQL", "ML"]
marks = [80, 70, 90, 85]

# 1. Line Plot
plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Line Plot")

# 2. Bar Plot
plt.subplot(2, 2, 2)
plt.bar(subjects, marks)
plt.title("Bar Plot")

# 3. Scatter Plot
plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title("Scatter Plot")

# 4. Histogram
plt.subplot(2, 2, 4)
plt.hist(marks, bins=5)
plt.title("Histogram")

plt.tight_layout()
plt.show()
```

Task - India flag using Matplotlib

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle
import numpy as np

fig, ax = plt.subplots(figsize=(10, 6))

# 🇮🇳 Indian Flag - 3 stripes
ax.add_patch(Rectangle((0, 2), 10, 2, color="orange"))
ax.add_patch(Rectangle((0, 0), 10, 2, color="white"))
ax.add_patch(Rectangle((0, -2), 10, 2, color="green"))

# Ashoka Chakra
center_x = 5
center_y = 1
radius = 0.75

# Circle
ax.add_patch(
    Circle((center_x, center_y),
           radius,
           fill=False,
           color="blue",
           linewidth=3)
)

# 24 spokes
for angle in np.arange(0, 360, 15):
    x = center_x + radius * np.cos(np.radians(angle))
    y = center_y + radius * np.sin(np.radians(angle))

    ax.plot(
        [center_x, x],
        [center_y, y],
        color="blue",
        linewidth=1
    )

# Settings
ax.set_xlim(0, 10)
ax.set_ylim(-2, 4)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```
