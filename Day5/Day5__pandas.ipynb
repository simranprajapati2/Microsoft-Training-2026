Task-Non-Linear Regression using sklearn,





```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split

salary['Years of Experience'] = salary['Years of Experience'].fillna(salary['Years of Experience'].median())

# Define x (features) and y (target)
x = salary[['Years of Experience']]
y = salary['Annual Salary']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
poly = PolynomialFeatures(degree=2)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train_poly, y_train)
```

```python
y_pred = model.predict(x_test_poly)

print(y_pred)
```

```python
x_curve = x.sort_values(by='Years of Experience')

x_curve_poly = poly.transform(x_curve)

y_curve_pred = model.predict(x_curve_poly)

plt.scatter(x, y)
plt.plot(x_curve, y_curve_pred)

plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Non-Linear Regression")

plt.show()
```
