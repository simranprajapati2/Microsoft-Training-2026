Object oriented programming


```python
#constructor
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Simran")
print(s1.name)
```

```python
#methon
class Student:
    def greet(self):
        print("Hello")

s1 = Student()
s1.greet()
```

```python
#encapsulation
class Student:
    def __init__(self):
        self.__marks = 90

    def show(self):
        print(self.__marks)

s = Student()
s.show()
```

```python
#inheritence
class Parent:
    def show(self):
        print("Parent Class")

class Child(Parent):
    pass

c = Child()
c.show()
```

```python
#polymorphism
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
#abstraction
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark")

d = Dog()
d.sound()
```

Python Collections Module

```python
from collections import Counter

text = "apple"

count = Counter(text)
print(count)
```

```python
from collections import namedtuple

Student = namedtuple("Student", ["name", "age"])

s = Student("Simran", 22)

print(s.name)
print(s.age)
```

```python
from collections import deque

d = deque([1, 2, 3])
d.append(4)
d.appendleft(0)

print(d)
```

```python
from collections import defaultdict

d = defaultdict(int)
d["a"] += 1

print(d)
```

open() → Opens a file.
Read ("r") → Reads data from a file.
Write ("w") → Writes data and overwrites existing content.
Append ("a") → Adds data to the end of a file.
close() → Closes the file and releases resources.
with open() → Opens a file and automatically closes it after the block finishes. It is the recommended way to work with files.

file = open("demo.txt", "a")
file.write("\nWelcome")
file.close()

with open("demo.txt", "r") as file:
