Dictionary Problem
d={'A':1,'B':2}
swap and reverse operation






```python
#swap value

d = {'A':1,'B':2}

temp = d['A']
d['A'] = d['B']
d['B'] = temp

print(d)
```

```python
#reverse order

d = {'A':1,'B':2,'c':3}

items = list(d.items())

items.reverse()

rev = dict(items)

print(rev)
```

d={'A':1,'B':3,'C':1}
print{1:A/c, B:3}



```python
d={'A':1,'B':3,'C':1}
print (d)

rev = {}

for k, v in d.items():
    if v not in rev:
        rev[v] = []
    rev[v].append(k)

print(rev)
```

Break vs Pass vs Continue

```python
# break – Stop the Loop/immediately terminates the loop.
for i in range(1,5):
  if i ==3:
    break
  print(i)
```

```python
#continue – Skip Current Iteration and  moves to the next one.
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
#pass – Do Nothing
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

Frequency count list

```python
lst = [1, 1, 2, 3, 2, 1, 4, 2,4]

print("1 element frequency:",lst.count(1))
print("2 element frequency:",lst.count(2))
print("3 element frequency:",lst.count(3))
print("4 element frequency:",lst.count(4))
```
